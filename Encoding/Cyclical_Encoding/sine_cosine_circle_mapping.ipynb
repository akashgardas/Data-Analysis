```python
import math
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def draw_sine_cosine_mapping(no_of_values=20):
    gradiants = [i*(2/no_of_values) for i in range(no_of_values)]
    print(f'Gradiants: \n{gradiants}')

    sin_values = [math.sin(i * math.pi) for i in gradiants]
    cos_values = [math.cos(i * math.pi) for i in gradiants]

    sns.scatterplot(sin_values, label='sine')
    sns.scatterplot(cos_values, color='r', label='cosine')
    plt.title('Sine & Cosine Curves')
    plt.legend()
    plt.show()

    data = {
        'sine': sin_values,
        'cosine': cos_values
    }
    
    df = pd.DataFrame(data)
    print(f'DataFrame: \n{df.head()}')

    ## sine vs cosine mapping - circle geometry
    plt.figure(figsize=(6, 6))
    # Scatter plot to show the circle
    sns.scatterplot(df, x='sine', y='cosine')
    plt.title('Sine VS Cosine Mapping - Circle')
    plt.grid(True)
    plt.show()
```

```python
draw_sine_cosine_mapping(50)
```
